## Custom functions

```python
def generate_slurm_array(cmds, output_file, simultaneous_jobs = 4):
    working_dir = os.getcwd()
    total_jobs = len(cmds)
    array_header = ['#!/usr/bin/bash',
                  f'#SBATCH --array=1-{total_jobs}%{simultaneous_jobs}',
                   '#SBATCH --job-name=modiscolite',
                   '#SBATCH --ntasks=1',
                   '#SBATCH --cpus-per-task=70',
                   '#SBATCH --mem=100gb',
                   '#SBATCH --time=48:00:00',
                   '#SBATCH --output=slurm_%j.log',
                  'source /home/mw2098/.bashrc',
                  'conda deactivate',
                   'ml meme',
                  'conda activate bpreveal_404',
                  f'cd {working_dir}']
    # array_cmds = array_header
    for i,cmd in enumerate(cmds):
        array_cmd = ["if [[ ${{SLURM_ARRAY_TASK_ID}} == {0:d} ]] ; then\n".format(i+1), 
                    "    {0:s}\n".format(cmd),
                    "fi\n\n"]
        array_header = array_header + array_cmd 
        
    with open(output_file, mode='wt') as slurm:
        slurm.write('\n'.join(array_header))
        slurm.write('\n')
```

# Run TF-MoDISco

```python
array_cmds=[]
for model_name,model_info in modeling_design_dict.items():
    tasks = list(model_info['cov'].keys())
    for fold in ['fold1', 'fold2', 'fold3']:
        for task in tasks:
            for mode in ['counts', 'profile']:
                
                #Run counts TF-MoDISco
                shap_to_npz = [f'{python_path} {bpreveal_path}/src/shapToNumpy.py \\',
                                      f'--h5 shap/{model_name}_{fold}_{task}_{mode}.h5 \\',
                                      f'--seqs npz/{model_name}_{fold}_{task}_{mode}_seqs \\',
                                      f'--scores npz/{model_name}_{fold}_{task}_{mode}_scores'] 
                modisco = [f'mkdir -p modiscolite/{model_name}_{fold}_{task}_{mode}',
                                  f'modisco motifs \\',
                                  f'-s npz/{model_name}_{fold}_{task}_{mode}_seqs.npy \\',
                                  f'-a npz/{model_name}_{fold}_{task}_{mode}_scores.npy \\',
                                  f'-n 10000 \\',
                                  f'-w {output_length} \\',
                                  f'--output modiscolite/{model_name}_{fold}_{task}_{mode}/modisco.h5']
                modisco_report = [f'modisco report -i modiscolite/{model_name}_{fold}_{task}_{mode}/modisco.h5 \\',
                                         f'-o modiscolite/{model_name}_{fold}_{task}_{mode} \\',
                                         f'-m {meme_db}']

                #Write to a job array
                cmds_str = '\n'.join(shap_to_npz + modisco + modisco_report)
                array_cmd = [cmds_str]
                array_cmds += array_cmd

generate_slurm_array(cmds = array_cmds, output_file = 'scripts/modiscolite_across_all_models.slurm', simultaneous_jobs=15)
print('sbatch scripts/modiscolite_across_all_models.slurm')
```

When processing the tfmodiscolite `modisco.h5` the internal code is as follows: 

modisco.h5: metacluster --> pattern_id -->
+ contrib_scores: CWM represented as a [30 x 4] array
+ hypothetical contribs: hypothetical CWM represented as a [30 x 4] array
+ sequence: PPM represented as a [30 x 4] array
+ seqlets -->
    + sequence, contrib_scores, hypothetical_contribs in a [n_seqlets x 30 x 4] array 
    + is_revcomp, n_seqlets, start, end, example_idx (peak index) in a [n_seqlets] array

# CWM scan motifs

For these representations, we need to CWM-scan motifs based on our mapping criteria and seqlet allocations. 

Map the motifs based off of sufficient contribution and contribution pattern adherence. 

```python
array_cmds=[]
for k in ['bunina_atac_day0_across_combined', 'bunina_atac_day4_across_combined', 'bunina_atac_day8_across_combined', 'bunina_atac_day12_across_combined']:

    short_prefix = k.replace('_fold1', '').replace('_fold2', '').replace('_fold3', '')
    v = modeling_design_dict[short_prefix]
    tasks = list(v['cov'].keys())
    
    for task in tasks:
        for mode in ['counts', 'profile']:

            # m_df = all_ids_df[(all_ids_df.model_name==k) & \
            #                     (all_ids_df.task==task) & \
            #                     (all_ids_df.contrib_type==mode)]
            
            # patterns_dict = [{'metacluster-name': row.metacluster, 
            #                   'pattern-name': row.pattern, 
            #                   'short-name': row.motif} for i, row in m_df.iterrows()]
            # if len(patterns_dict)>0:
            
            motif_scan_dict = {
                'seqlet-cutoff-settings': {
                    'quantile-json': f'json/motifSeqletCutoffs_{k}_{task}_{mode}_quantiles.json', 
                    'modisco-h5': f'modiscolite/{k}_fold1_{task}_{mode}/modisco.h5',
                    'modisco-contrib-h5': f'shap/{k}_fold1_{task}_{mode}.h5',
                    'patterns': 'all', 
                    'seq-match-quantile': 0.01,
                    'contrib-match-quantile': 0.2, 
                    'contrib-magnitude-quantile': 0.01,
                    'trim-threshold': 0.3,
                    'trim-padding': 1,
                    'background-probs': [.29, .21, .21, .29], #dm6 GC content
                    'seqlets-tsv': f'modiscolite/{k}_fold1_{task}_{mode}/seqlets.tsv',
                    'verbosity': 'WARNING'
                },
                'scan-settings': {
                    'scan-contrib-h5': f'shap/{k}_fold1_{task}_{mode}.h5',
                    'hits-tsv': f'modiscolite/{k}_fold1_{task}_{mode}/hits.tsv',
                    'num-threads': 68
                }, 
                'verbosity': 'WARNING'
            }
            motif_scan_json = json.dumps(motif_scan_dict, indent=4)
            motif_scan_file = f'json/prepareMotifSeqletCutoffs_{k}_{task}_{mode}.json'        
            with open(motif_scan_file, 'w') as outfile:
                outfile.write(motif_scan_json) 

            motif_scan = [f'{python_path} {bpreveal_path}/src/motifScan.py {motif_scan_file}']

            #Write to a job array
            cmds_str = '\n'.join(motif_scan)
            array_cmd = [cmds_str]
            array_cmds += array_cmd

generate_slurm_array(cmds = array_cmds, output_file = 'scripts/cwmscan_across_selected_models.slurm', simultaneous_jobs=15)
print('sbatch scripts/cwmscan_across_selected_models.slurm')
```
